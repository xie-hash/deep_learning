```python

import os
tran_compose = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize([0.3,0.3,0.3],[0.5,0.5,0.5]),
    transforms.Resize([512,512]),
])

dir_path = r"C:\Users\谢\Desktop\vscodepj\深度学习\hymenoptera_data\hymenoptera_data\train\ants"
img_path_list = os.listdir(dir_path)[0:10]
record_compose = SummaryWriter("logs_compose")
for i in enumerate(img_path_list):
    img = Image.open(os.path.join(dir_path, i[1]))
    img_tensor = tran_compose(img)
    record_compose.add_image("img_compose_2", img_tensor, i[0])
record_compose.close()
```

```python
import os
tran_compose = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize([0.3,0.3,0.3],[0.5,0.5,0.5]),
    transforms.Resize([512,512]),
    transforms.RandomCrop([128,128]),
])

dir_path = r"C:\Users\谢\Desktop\vscodepj\深度学习\hymenoptera_data\hymenoptera_data\train\ants"
img_path_list = os.listdir(dir_path)[1]
img_path_list = [img_path_list for i in range(10)]
record_compose = SummaryWriter("logs_crop")
for i in enumerate(img_path_list):
    print(i)
    img = Image.open(os.path.join(dir_path, i[1]))
    img_tensor = tran_compose(img)
    record_compose.add_image("img_crop_2", img_tensor, i[0])
record_compose.close()
```
