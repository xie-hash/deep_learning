```python
from torch import nn
import torch
from torch.utils.tensorboard import SummaryWriter
import torch.utils.tensorboard
```

```python
class Seq_nn(nn.Module):
    def __init__(self):
        super(Seq_nn, self).__init__()
        self.module = nn.Sequential(
            nn.Conv2d(3,32,5,stride=1,padding=2,dilation=1,bias=True),
            nn.MaxPool2d(kernel_size=2,padding=0,stride=2),
            nn.Conv2d(in_channels=32,out_channels=32,kernel_size=5,stride=1,padding=2,dilation=1,bias=True),
            nn.MaxPool2d(kernel_size=2,padding=0,stride=2), 
            nn.Conv2d(in_channels=32,out_channels=64,kernel_size=5,stride=1,padding=2,dilation=1,bias=True),
            nn.MaxPool2d(kernel_size=2,padding=0,stride=2),
            nn.Flatten(),
            nn.Linear(in_features=1024,out_features=64,bias=True),
            nn.Linear(in_features=64,out_features=10,bias=True)
        )
    def forward(self, x):
        return self.module(x)


        
```

```python
input = torch.ones((64,3,32,32))
print(input.shape)
seq = Seq_nn()
output = seq(input)
print(output.shape)

writer = SummaryWriter(log_dir="logs_seq")
writer.add_graph(seq,input)
writer.close()
```

```python
input2 = torch.zeros((64,3,32,32))
print(input2.shape)
seq = Seq_nn()
output2 = seq(input2)
output2 = torch.reshape(output2, (64,10,1,1))
output = torch.reshape(output, (64,10,1,1))
print(output2.shape)
print(output.shape)
```

```python
loss = nn.L1Loss()
loss_value = loss(output2, output)
print(loss_value)
```

```python
loss_mse = nn.MSELoss()
loss_mse_value = loss_mse(output2, output)
print(loss_mse_value)
```
