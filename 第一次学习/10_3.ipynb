```python
import torch
from torch.utils.data import Dataset, DataLoader
from PIL import Image
class Data_Creat(Dataset):
    def __init__(self, root,root_dir):
        self.root = root
        self.root_dir = root_dir
        self.img_list = os.listdir(os.path.join(self.root, self.root_dir))

    def __getitem__(self, index):
        img_name = self.img_list[index]
        img_path = os.path.join(self.root, self.root_dir, img_name)
        img = Image.open(img_path)
        return img, self.root_dir
    def __len__(self):
        return len(self.img_list)

```

```python
root = r"C:\Users\谢\Desktop\vscodepj\深度学习\hymenoptera_data\hymenoptera_data\train"
root_dir = "ants"
dataset = Data_Creat(root, root_dir)
```
