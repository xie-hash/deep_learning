```python
import torch
import torchvision
from torch import nn
```

```python
dataset1 = torchvision.datasets.CIFAR10(root='C:\\Users\\谢\\Desktop\\vscodepj\\深度学习\\第三次学习\\dataset', train=True, download=True,transform=torchvision.transforms.ToTensor())

dataloader1 = torch.utils.data.DataLoader(dataset1, batch_size=64, shuffle=True, num_workers=2)
print(dataset1[0][0])
```

```python
class xie_nn(nn.Module):
    def __init__(self):
        super(xie_nn, self).__init__()
        self.conv1 = nn.Conv2d(3, 6, stride=1, padding=0, kernel_size=3)
    
    def forward(self,input):
        output = self.conv1(input)
        return output
    

```

```python
xie = xie_nn()
```

```python
print(xie.forward(dataset1[1][0]).shape)
print(dataset1[1][0].shape)
```

```python
from torch.utils import tensorboard
imgs_writer = tensorboard.SummaryWriter('logs')

```

```python
step = 0
for data in dataloader1:
    tensor,label = data
    output = xie.forward(tensor)
    output = torch.reshape(output,(-1,3,30,30))
    imgs_writer.add_images('input',tensor,global_step=step)
    imgs_writer.add_images('output',output,global_step=step)
    step += 1
    
```
