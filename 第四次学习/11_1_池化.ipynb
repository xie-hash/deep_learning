```python
#池化 maxpool
import torch
from torch.nn.functional import max_pool2d
```

```python
input = torch.Tensor([[1,2,0,3,1],
                      [0,1,2,3,1],
                      [1,2,1,0,0],
                      [5,2,3,1,1],
                      [2,1,0,1,1]])
kenel_size = 3
input = torch.reshape(input,(-1,1,5,5))
output = max_pool2d(input,kenel_size,ceil_mode=True,stride=1)
print(output)
```

```python
import torchvision
from torchvision import datasets,transforms
from torch.utils.data import DataLoader
dataset = torchvision.datasets.CIFAR10(r"C:\Users\谢\Desktop\vscodepj\深度学习\第三次学习\dataset",download=True,train=True,transform=transforms.ToTensor())
dataloader = DataLoader(dataset,batch_size=64,shuffle=True)
```

```python
from torch import nn
class xie_nn(nn.Module):
    def __init__(self):
        super(xie_nn,self).__init__()
        self.conver1 = nn.Conv2d(3,6,stride=1,padding=1,kernel_size=4)
        self.conver2 = nn.MaxPool2d(kernel_size=2,stride=2,ceil_mode=True)
    
    def forward(self,x):
        x = self.conver1(x)
        x = self.conver2(x)
        return x
```

```python
from torch.utils import tensorboard
imgs_writer = tensorboard.SummaryWriter('logs_2')
```

```python
step = 0
xie = xie_nn()
for data in dataloader:
    tensor,label = data
    output = xie(tensor)
    output = torch.reshape(output,(-1,3,16,16))
    imgs_writer.add_images('input',tensor,global_step=step)
    imgs_writer.add_images('output',output,global_step=step)
    step += 1

    
    
```
