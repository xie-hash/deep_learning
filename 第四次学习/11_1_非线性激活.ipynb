```python
import torch
from torch import nn
from torch.nn import functional as F
input = torch.Tensor([[1,-1,0,3,1],
                      [0,-3,2,3,1],
                      [1,2,1,0,0],
                      [5,2,3,1,1],
                      [2,1,0,1,1]])

input = torch.reshape(input,(-1,1,5,5))
output = F.relu(input)
print(output)
print(input)
output2 = F.sigmoid(input)
print(output2)
```
