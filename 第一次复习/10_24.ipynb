```python
from torch.utils.data import Dataset
import os
from PIL import Image
from torchvision import Transform
class HymenopteraDataset(Dataset):
    def __init__(self,root,split,transform=None):
        self.root = root
        if split=="ants":
            self.label = 0
        else:
            self.label = 1
        self.name = split
        self.file_list_path = os.path.join(root,split)
        self.file_list = os.find_dir(self.file_list_path)
        self.trans = transform
        
    def __getitme__(self,idex):
        file_name = self.file_list[idex]
        file_path = os.path.join(self.file_list_path,file_name)
        img = Image.open(file_path)
        img_trans = self.trans(img)
        return img_trans,self,label
    
    def __len__(self):
        return len(self.file_path)
    
        
        
```
